## 1. Question

*The research question and the decision it supports.*
Content Refresh Prioritization Using Historical Search Performance Signals: A Baseline and Machine Learning Approach
This project investigates whether historical search-performance signals can be used to prioritize content refresh opportunities. Using the FlyRank ML Internship dataset, I analyzed content age, traffic trends, engagement metrics, and search-performance indicators to identify pages that may benefit from editorial review. A rule-based baseline was developed and compared against a machine-learning model using the same evaluation framework. Validation included leakage checks, grouped evaluation by client, and error analysis to ensure honest interpretation of results. The resulting action playbook provides ranked recommendations intended for human decision-support rather than automated content modification.
Content performance often changes over time due to shifts in search demand, competition, and content freshness. Organizations with large content portfolios cannot manually review every page on a regular basis.

The objective of this project was to develop a decision-support system that helps prioritize content for review using historical search-performance signals. The goal was not to predict future SEO outcomes with certainty, but rather to identify content that may warrant human attention based on observed patterns in historical data.

## 5. Limitations

*What this work cannot claim.*
This project is observational.

The model identifies patterns associated with refresh opportunities but does not establish causal relationships.

The analysis does not measure actual post-refresh performance.

Seasonality, business priorities, and content quality were not directly modeled.

Recommendations should be interpreted as decision-support signals rather than guaranteed outcomes.

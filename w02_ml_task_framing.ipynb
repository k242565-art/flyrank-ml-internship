## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
A fixed rule (like "flag any page with days_since_last_update >= 180 and impressions_90d >= 500") only looks at one or two signals at a time and can't weigh how multiple weak signals combine. The starter pipeline's own baseline rule scored 0.240 precision@50, while the random forest — which learns how to combine impressions, position, freshness, and CTR together — hit 0.740. That's not a marginal gain; it's roughly 3x more correct picks in the same top-50 slots, using the exact same underlying signals. The rule can't adapt weightings automatically; the model can.
